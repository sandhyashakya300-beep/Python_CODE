# BASIC PYTHON CODE RELATED TO FUNCTION

```python
# Write a function to calculate and return the square of a given number.
def square_of_number(num):
    return num ** 2

number = int(input("Enter a number to calculate its square: "))
result = square_of_number(number)

print(f"the square of {number} is {result}.")
```

```python
# Create a function that takes two numbers as parameters and returns their sum.
def sum_of_numbers(num1, num2):
    return num1 + num2 

num1 = int(input("Enter the first number: "))
num2 = int(input("Enter the second number: "))
sum_of_result = sum_of_numbers(num1, num2)

print(f" The sum of {num1} and {num2} is {sum_of_result}.")
```

```python
#Polymorphism in functions 
# Write a function multiply the multiplies two numbers , but can also accept and multiply string 
def multiply(a , b ):
    return a * b

a = input("Enter the first  input: ")
b = int(input("Enter the second input : "))

result = multiply(a , b )

print(f"The multiplication of {a } and {b} is {result}")
```

```python
# Create a function that return both the area and circumference of a circle given its radius.
def circle_A_C(radius):
    area = 3.14 * radius **2
    circumference = 2 * 3.14 * radius
    return area , circumference

radius = float(input("Enter the radius of the cicle: "))

area , circumference = circle_A_C(radius)

print(f" The area of the circle with radius {radius} is {area} and the circumference is {circumference}  ")
```

```python
# Write a function that greets a user . if no name is provided , it should greet with a default name.
def greet_user(name = "User"):
    if not name:
        name = "User"
    return f"Hello, {name}!"

name = input("Enter the name of the user (or press Enter to use default):")
greeting = greet_user(name)

print(f"{greeting}",end=" ")
```

```python
# Create a lamda function to compute the cube of a number.
cube = lambda x: x ** 3

num = int(input("Enter the number to compute its cube: "))
result = cube(num)
print(f"The cube of {num} is {result}.")
```

```python
# Write a function that takes variable number of arguments and returns their sum.
def sum_of_args(*args):
    print(f"Arguments received: {args}")
    return sum(args)

result = sum_of_args(1, 2, 3, 4, 5)
print(f"The sum of the arguments is {result}.")
```

```python
# Create a function that accepts any number of keyword arguments and prints them in the format key : value.
def print_kwargs(**kwargs):
    for key, value in kwargs.items():
        print(f"{key}: {value}")

print_kwargs(name="Subham", age=33, city="Badaun")
```

```python
# write a generator function that yields the even numbers upto a specified limit.
def even_generator(limit):
    for i in range(2, limit + 1, 2):
        yield i 

for num in even_generator(30):
    print(num)
```

```python
# Create a recursion function to calculate the factorial of a number.
def factorial(n):
    if n == 0 or n == 1:
        return 1
    else:
        return n * factorial(n-1)


num = int(input("Enter the number you want to find factorial: "))
result = factorial(num)
print(result)
```
