```python
num1 = int(input("Enter first number: "))
if num1> 0:
    print(f"The number is positive: {num1}")
    if num1 % 2 == 0:
        print(f"The number is even: {num1}")
    else:        print(f"The number is odd: {num1}")
elif num1 < 0:
    print(f"The number is negative: {num1}")
    if num1 % 4 == 0:
        print(f"The number is divisible by 4: {num1}")
    else:        print(f"The number is not divisible by 4: {num1}")
elif num1>0:
    if num1 % 3 == 0:
        print(f"The number is divisible by 3: {num1}")
    else:        print(f"The number is not divisible by 3: {num1}") 
else:    print(f"The number is zero: {num1}")
```

```python
#H.W you have to check if the triangle is acute angle or obtuse angle or right angle triangle equilateral triangle

side1 = float(input("Enter the length of the first side of the triangle: "))
side2 = float(input("Enter the length of the second side of the triangle: "))   
side3 = float(input("Enter the length of the third side of the triangle: "))
if side1 == side2 == side3:
    print("The triangle is an equilateral triangle.")
elif side1 == side2 or side2 == side3 or side1 == side3:
    print("The triangle is an isosceles triangle.")
elif (side1**2 + side2**2 == side3**2) or (side1**2 + side3**2 == side2**2) or (side2**2 + side3**2 == side1**2):
    print("The triangle is a right angle triangle.")
elif (side1**2 + side2**2 < side3**2) or (side1**2 + side3**2 < side2**2) or (side2**2 + side3**2 < side1**2):
    print("The triangle is an obtuse angle triangle.")
else:   
    print("The triangle is an acute angle triangle.")
```

```python
marks = [85, 92, 78, 90, 88]
for mark in marks:
    if mark >= 90:
        print(f"{mark} is an A grade.")
    elif mark >= 80:
        print(f"{mark} is a B grade.")
    elif mark >= 70:
        print(f"{mark} is a C grade.")
    elif mark >= 60:
        print(f"{mark} is a D grade.")
    else:
        print(f"{mark} is an F grade.")
marks.append(95)
print(marks)
sum_marks = sum(marks)
average_marks = sum_marks / len(marks)  
print(f"Total marks: {sum_marks}")
print(f"Average marks: {average_marks}")    
```
